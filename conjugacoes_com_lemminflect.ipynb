```python
tense_mask = {
    "VB" : "Base",
    "VBD" : "Past_Simple",
    "VBG" : "Present_Participle",
    "VBN" : "Past_Participle",
    "VBZ" : "Third_Person_Present"
}
```

```python
import spacy
from lemminflect import getInflection

nlp = spacy.load("en_core_web_sm")

verb = "have"
doc = nlp(verb)

lemma = doc[0].lemma_

# Formas verbais disponíveis para determinadas categorias
for tag in ["VB", "VBD", "VBG", "VBN", "VBZ"]:
    forms = getInflection(lemma, tag=tag)
    print(f"{tense_mask[tag]}: {forms[0]}")
```
