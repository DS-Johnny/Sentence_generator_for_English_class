```python
from random import choice
```

```python
places = {
    "hospital" : "hospital",
    "delegacia" : "police station",
    "mercado" : "supermarket",
    "pista de skate" : "skate park"
}

areas = {
    "na minha cidade" : ("city", "in my"),
    "no meu bairro" : ("neighborhood", "in my"),
    "por aqui" : ("around here", "")
}

forms = ["q", "a", "n"]
```

```python
for k,v in places.items():
    print({f'{k}' : v})
```

```python
# place = choice([{f'{k}': v} for k,v in places.items()])
place = choice([i for i in places])

place, places[place]
```

```python
area = choice([i for i in areas])
area, areas[area]
```

```python
def gerar():
    place = choice([i for i in places])
    area = choice([i for i in areas])

        
    form = choice(forms)
    if form == "a":
        pt = f'Tem um {place} {area}'
        en = f'There is a {places[place]} {areas[area][1]} {areas[area][0]}'
    elif form == "q":
        pt = f'Tem um {place} {area}?'
        en = f'Is there a {places[place]} {areas[area][1]} {areas[area][0]}?'
    else:
        pt = f'Não tem um {place} {area}'
        en = f'There is not a {places[place]} {areas[area][1]} {areas[area][0]}'

    return pt,en
```

```python
gerar()
```
