```python
from faker import Faker
from random import choice, randint
fake = Faker()#locale='pt-br')

fake.first_name()

# fake.country()
```

```python
fake.first_name_male(), fake.first_name_female()
```

```python
def origem():
    verbos = {
        'eu' : 'sou',
        'você' : 'é',
        'ele' : 'é',
        'ela' : 'é',
        'nós' : 'somos',
        'vocês' : 'são',
        'eles' : 'são',
        'elas' : 'são'
    }
    sujeito = choice([key for key in verbos.keys()])
    tipo = choice(['afirmativa', 'interrogativa', 'negativa'])

    if tipo == 'afirmativa':
        return tipo, f'{sujeito} {verbos[sujeito]} do {fake.country()}'

    elif tipo == 'negativa':
        return tipo, f'{sujeito} não {verbos[sujeito]} do {fake.country()}'

    else:
        return tipo, f'{sujeito} {verbos[sujeito]} do {fake.country()}?'
```

```python
def name():
    possessives = {
        'meu nome' : 'é',
        'seu nome' : 'é',
        'o nome dele' : 'é',
        'o nome dela' : 'é',
        'nossos nomes' : 'são',
        'seus nomes' : 'são',
        'os nomes deles' : 'são'
    }

    sujeito = choice([key for key in possessives.keys()])
    verb = possessives[sujeito]
    tipo = choice(['afirmativa', 'interrogativa', 'negativa'])

    if tipo == 'afirmativa':
        if verb == 'são':
            return tipo, f'{sujeito} {verb} {fake.first_name()} and {fake.first_name()}'
        else:
            return tipo, f'{sujeito} {verb} {fake.first_name()}'

    elif tipo == 'negativa':
        if verb == 'são':
            return tipo, f'{sujeito} não {verb} {fake.first_name()} and {fake.first_name()}'
        else:
            return tipo, f'{sujeito} não {verb} {fake.first_name()}'

    else:
        if verb == 'são':
            return tipo, f'{sujeito} {verb} {fake.first_name()} and {fake.first_name()}?'
        else:
            return tipo, f'{sujeito} {verb} {fake.first_name()}?'
    
    
```

```python
def age():
    verbos = {
            'eu' : 'tenho',
            'você' : 'tem',
            'ele' : 'tem',
            'ela' : 'tem',
            'nós' : 'temos',
            'vocês' : 'tem',
            'eles' : 'tem',
            'elas' : 'tem'
        }


    sujeito = choice([key for key in verbos.keys()])
    verbo = verbos[sujeito]
    idade = randint(1,110)
    tipo = choice(['afirmativa', 'interrogativa', 'negativa'])
    if tipo == 'afirmativa':
        return tipo, f'{sujeito} {verbos[sujeito]} {idade} anos.'

    elif tipo == 'negativa':
        return tipo, f'{sujeito} não {verbos[sujeito]} {idade} anos.'

    else:
        return tipo, f'{sujeito} {verbos[sujeito]} {idade} anos?'
```

```python
def gerador():
    topics = {
    'origem' : origem,
    'nome' : name,
    'idade' : age
    }

    return topics[choice([key for key in topics.keys()])]()

gerador()
```
