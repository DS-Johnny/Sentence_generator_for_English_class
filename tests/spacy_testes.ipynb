```python
import spacy 

nlp = spacy.load("pt_core_news_lg")
```

```python
text = "Tem uma hotel em Nova Orleans"
```

```python
doc = nlp(text)
for i in doc:
    print(i)
```

```python
dados= doc.to_dict()
for k,v in dados.items():
    print(f'Key: {k:25} | Values: {type(v)}')
```

```python
texto = "O médico e a cientista analisaram o relatório da estrutura."
doc = nlp(texto)

for token in doc:
    # Filtra apenas para analisar substantivos (NOUN) e nomes próprios (PROPN)
    if token.pos_ in ["NOUN", "PROPN"]:
        # Obtém a lista do atributo Gender (retorna [] se não mapeado)
        genero_lista = token.morph.get("Gender")
        
        # Extrai o valor da lista se ela não estiver vazia
        genero = genero_lista[0] if genero_lista else "Não identificado"
        
        print(f"Substantivo: {token.text:<12} | Gênero: {genero}")
```
